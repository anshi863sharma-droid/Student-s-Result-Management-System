**Student Result Management App**

Project Overview

This project is developed using Python to manage student academic records and calculate their performance.

Objectives:

Store student names and marks.
Calculate total and percentage.
Assign grades based on performance.
Determine pass/fail status.
Display student results.



```python
students = []

print("================================")
print(" STUDENT RESULT MANAGEMENT APP ")
print("================================")
```

```python
n = int(input("Enter number of students: "))

for i in range(n):

    print("Student", i+1)

    name = input("Enter student name: ")

    marks = []

    for j in range(5):

        mark = float(input(f"Enter marks for subject {j+1}: "))

        marks.append(mark)

    students.append([name, marks])

print("Student records successfully stored!")
```

```python
for student in students:

    name = student[0]
    marks = student[1]

    total = sum(marks)
    percentage = total / 5

    print("Student Name:", name)
    print("Total Marks:", total)
    print("Percentage:", percentage, "%")
```

```python
for student in students:

    name = student[0]
    marks = student[1]

    percentage = sum(marks) / 5

    if percentage >= 90:
        grade = "A+"

    elif percentage >= 80:
        grade = "A"

    elif percentage >= 70:
        grade = "B"

    elif percentage >= 60:
        grade = "C"

    elif percentage >= 50:
        grade = "D"

    elif percentage >= 40:
        grade = "E"

    else:
        grade = "F"

    print(name, "Grade:", grade)
```

```python
for student in students:

    name = student[0]
    marks = student[1]

    if min(marks) >= 33:
        result = "PASS"

    else:
        result = "FAIL"

    print(name, "Result:", result)
```

```python
print("== FINAL STUDENT REPORT ==")

for student in students:

    name = student[0]
    marks = student[1]

    total = sum(marks)
    percentage = total / 5

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    elif percentage >= 40:
        grade = "E"
    else:
        grade = "F"

    if min(marks) >= 33:
        result = "PASS"
    else:
        result = "FAIL"

    print("Student:", name)
    print("Marks:", marks)
    print("Total:", total, "/ 500")
    print("Percentage:", percentage, "%")
    print("Grade:", grade)
    print("Final Result:", result)

    print("--------------------------------")
```
